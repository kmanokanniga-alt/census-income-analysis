# Stage 2 – Data Cleaning and Pre-processing

## Objective

The objective of this stage is to improve data quality by handling missing values, removing duplicate records, checking data types, and preparing the dataset for further exploratory analysis and visualization.

### Data Cleaning Summary

The dataset was cleaned and prepared for further analysis through a systematic data preprocessing process.

* Missing values in categorical columns were handled using the mode.
* Duplicate records were identified and removed.
* Data types of all columns were checked and validated.
* Unnecessary whitespace was removed from categorical values.
* Potential outliers in numerical variables were identified.
* Skewness in numerical variables was evaluated.
* Log transformation was applied to highly skewed financial variables such as capital-gain and capital-loss.
* Derived features such as Age Group and Work Hours Category were created for further analysis.

The original dataset contained **32,561 records and 15 columns**. During preprocessing, **24 duplicate records were identified and removed**, resulting in **32,537 records** for analysis.


```python
# Create a copy for data cleaning

df_clean = df.copy()

# Fill missing values in categorical columns with their mode
categorical_columns = ['workclass', 'occupation', 'native-country']

for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].mode()[0])

# Check missing values after treatment
print("Missing values after treatment:")
print(df_clean.isnull().sum())
```

```python
# Remove duplicate records

duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

duplicates_after = df_clean.duplicated().sum()

print("Duplicate records before treatment:", duplicates_before)
print("Duplicate records after treatment:", duplicates_after)
print("Shape after removing duplicates:", df_clean.shape)
```

```python
# Check data types after cleaning

print("Data types of each column:")
print(df_clean.dtypes)
```

```python
# Remove leading and trailing spaces from categorical columns

text_columns = df_clean.select_dtypes(include='object').columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Whitespace cleaning completed successfully.")
```

```python
# Detect outliers using the IQR method

numerical_columns = [
    'age',
    'fnlwgt',
    'education-num',
    'capital-gain',
    'capital-loss',
    'hours-per-week'
]

outlier_summary = {}

for column in numerical_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df_clean[column] < lower_bound) |
                (df_clean[column] > upper_bound)).sum()

    outlier_summary[column] = outliers

print("Outlier count for each numerical column:")
for column, count in outlier_summary.items():
    print(f"{column}: {count}")
```

## Outlier Treatment

Outliers were identified using the Interquartile Range (IQR) method. Several numerical columns contain potential outliers, particularly capital-gain, capital-loss, and hours-per-week.

These observations were retained because they may represent genuine variations in income, working hours, and financial characteristics rather than data-entry errors. Therefore, no records were removed solely based on the IQR outlier detection.

```python
# Check skewness of numerical columns

skewness = df_clean[numerical_columns].skew()

print("Skewness of numerical columns:")
print(skewness)
```

```python
# Apply log transformation to highly skewed financial variables

df_clean['capital-gain-log'] = np.log1p(df_clean['capital-gain'])
df_clean['capital-loss-log'] = np.log1p(df_clean['capital-loss'])

print("Log transformation applied successfully.")

print("\nSkewness after transformation:")
print("capital-gain-log:", df_clean['capital-gain-log'].skew())
print("capital-loss-log:", df_clean['capital-loss-log'].skew())
```

## Skewness Treatment

The skewness of numerical variables was evaluated before transformation. The capital-gain and capital-loss variables showed high positive skewness.

A log1p transformation was applied to reduce the effect of extreme values while preserving zero values. The transformation reduced the skewness of capital-gain from 11.95 to approximately 3.09 and capital-loss from 4.59 to approximately 4.31.

The transformed variables were retained as additional features for analysis, while the original variables were preserved.

```python
# Create meaningful derived features

# Age group
df_clean['age_group'] = pd.cut(
    df_clean['age'],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
)

# Work hours category
df_clean['work_hours_category'] = pd.cut(
    df_clean['hours-per-week'],
    bins=[0, 20, 40, 60, 100],
    labels=['Part-time', 'Full-time', 'Overtime', 'Very High']
)

print("Feature transformation completed successfully.")

print("\nAge Group:")
print(df_clean['age_group'].value_counts().sort_index())

print("\nWork Hours Category:")
print(df_clean['work_hours_category'].value_counts().sort_index())
```

```python
# Check the distribution of income categories

print("Income Category Distribution:")
print(df_clean['income'].value_counts())
```

```python
# Age Group vs Income Category Analysis

# Create age groups
df_clean['age_group'] = pd.cut(
    df_clean['age'],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
)

# Calculate percentage distribution
age_income = pd.crosstab(
    df_clean['age_group'],
    df_clean['income'],
    normalize='index'
) * 100

# Plot
ax = age_income.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 6),
    width=0.75
)

plt.title('Income Category Distribution Across Age Groups', fontsize=16, fontweight='bold')
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Percentage of Individuals (%)', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Income Category', bbox_to_anchor=(1.02, 1), loc='upper left')

# Add percentage labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', label_type='center', fontsize=9)

plt.tight_layout()
plt.show()
```

 ### Insight
The 18–25 age group has the highest proportion of individuals earning ≤50K (98.2%), while the 46–55 age group has the highest proportion earning >50K (39.9%).

### Interpretation
Higher income is more common among middle-aged individuals, especially those aged 46–55, while younger individuals are mostly in the ≤50K income category.

```python
# Age, Education and Income - Bubble Chart

bubble_data = (
    df_clean.groupby(['education', 'income'])
    .agg(
        avg_age=('age', 'mean'),
        people_count=('age', 'count')
    )
    .reset_index()
)

plt.figure(figsize=(12, 8))

for income_category in bubble_data['income'].unique():
    data = bubble_data[bubble_data['income'] == income_category]

    plt.scatter(
        data['avg_age'],
        data['education'],
        s=data['people_count'] / 5,
        alpha=0.6,
        label=income_category
    )

plt.title(
    'Age, Education and Income Distribution',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Average Age')
plt.ylabel('Education Level')
plt.legend(title='Income Category')

plt.tight_layout()
plt.show()
```

### Insight

The chart shows that higher-income individuals generally have higher average ages across education levels. Higher education levels such as Masters, Doctorate, and Prof-school show a greater presence in the >50K income category.

### Interpretation

Age, education level, and income are related, with higher education levels generally showing a higher representation of individuals earning >50K.

```python
#  High-Income Percentage by Age Group

age_group_temp = pd.cut(
    df_clean['age'],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
)

age_income = (
    df_clean.assign(age_group_temp=age_group_temp)
    .groupby('age_group_temp', observed=True)['income']
    .apply(lambda x: (x == '>50K').mean() * 100)
    .reset_index(name='high_income_percentage')
)

plt.figure(figsize=(10, 6))

plt.step(
    age_income['age_group_temp'].astype(str),
    age_income['high_income_percentage'],
    where='mid',
    linewidth=3
)

plt.scatter(
    age_income['age_group_temp'].astype(str),
    age_income['high_income_percentage'],
    s=100
)

for x, y in zip(
    age_income['age_group_temp'].astype(str),
    age_income['high_income_percentage']
):
    plt.text(
        x, y + 1,
        f'{y:.1f}%',
        ha='center',
        fontweight='bold'
    )

plt.title(
    'High-Income Percentage Across Age Groups',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Age Group')
plt.ylabel('Individuals Earning >50K (%)')

plt.tight_layout()
plt.show()
```

### Insight

The 46–55 age group has the highest proportion of individuals earning >50K (39.8%), while the 18–25 group has the lowest (1.9%).

### Interpretation

Higher income is more common among middle-aged individuals, with the highest proportion observed in the 46–55 age group.

```python
#  Age vs Weekly Working Hours – Data Concentration

plt.figure(figsize=(10, 6))

plt.hexbin(
    df_clean['age'],
    df_clean['hours-per-week'],
    gridsize=35,
    mincnt=1,
    bins='log'
)

plt.colorbar(label='Number of Individuals (log scale)')

plt.title(
    'Age vs Weekly Working Hours – Data Concentration',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Age')
plt.ylabel('Hours per Week')

plt.tight_layout()
plt.show()
```

### Insight

Most individuals are concentrated around 30–50 working hours per week, especially among ages 25–60.

### Interpretation

The chart shows that standard working hours are more common across the working-age population, while very low or very high working hours are less concentrated.
